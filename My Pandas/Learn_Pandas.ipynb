### 6. Data Transformation and Analysis

```python
# sorting data

df.sort_values(by='study_hours')   # sort the data by the 'study_hours' column in ascending order. It will return a new DataFrame sorted by the 'study_hours' column. The original DataFrame will not be modified.
```

```python
# accending order

df.sort_values(by='study_hours', ascending=True)   # sort the data by the 'study_hours' column in ascending order. It will return a new DataFrame sorted by the 'study_hours' column. The original DataFrame will not be modified.
```

```python
# decending order

df.sort_values(by='study_hours', ascending=False)   # sort the data by the 'study_hours' column in descending order. It will return a new DataFrame sorted by the 'study_hours' column. The original DataFrame will not be modified.
```

```python
# sorting by multiple columns

df.sort_values(by=['study_hours', 'previous_grade'], ascending=[True, False])   # sort the data by the 'study_hours' column in ascending order and then by the 'previous_grade' column in descending order. It will return a new DataFrame sorted by the 'study_hours' column and then by the 'previous_grade' column. The original DataFrame will not be modified.
```

```python
# value counts

df['tutoring'].value_counts()   # show the count of unique values in the 'tutoring' column. It will return a Series with the unique values as the index and the count of each unique value as the values.
```

```python
df['pass'].value_counts()   # show the count of unique values in the 'pass' column. It will return a Series with the unique values as the index and the count of each unique value as the values.
```

```python
# grouping data

df.groupby('tutoring')['study_hours'].mean()   # group the data by the 'tutoring' column and calculate the mean of the 'study_hours' column for each group. It will return a Series with the unique values in the 'tutoring' column as the index and the mean of the 'study_hours' column for each group as the values.
```

```python
# Feature Engineering

df['total_study_time'] = df['study_hours'] + df['attendance']   # create a new column 'total_study_time' by adding the 'study_hours' and 'attendance' columns. It will return a new Series with the total study time for each row. The original DataFrame will be modified to include the new column.
```

```python
df['good_students'] = 'No'  # create a new column 'good_students' and set the default value to 'No'. It will return a new Series with the value 'No' for each row. The original DataFrame will be modified to include the new column.
df.loc[(df['study_hours'] > 30 ) & (df['attendance'] > 70), 'good_students'] = 'Yes'    # update the 'good_students' column to 'Yes' for the rows where the 'study_hours' is greater than 30 and the 'attendance' is greater than 70. It will return a new Series with the updated values for the 'good_students' column. The original DataFrame will be modified to include the updated values in the 'good_students' column.
```

```python
df['good_students'].value_counts()  # show the count of unique values in the 'good_students' column. It will return a Series with the unique values as the index and the count of each unique value as the values.
```

### Merging DataFrames

```python
# create 2 data frames

students = pd.DataFrame({
    'student_id': [101, 102, 103, 104],
    'name': ['Ramesh', 'Suresh', 'Mahesh', 'Naresh'],
    'city': ['Hyderabad', 'Vijayawada', 'Guntur', 'Nellore']
})

marks = pd.DataFrame({
    'student_id': [101, 102, 102, 105],
    'subject': ['Maths', 'Maths', 'Science', 'Maths'],
    'marks': [85, 78, 88, 90]
})
```

```python
students.head()
```

```python
marks.head()
```

```python
# inner (common)

inner_merged_df = pd.merge(students, marks, on= 'student_id', how= 'inner')
```

```python
inner_merged_df
```

```python
# outer (union)

outer_merged_df = pd.merge(students, marks, on= 'student_id', how= 'outer')
```

```python
outer_merged_df
```
