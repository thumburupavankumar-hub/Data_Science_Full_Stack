```python
"""

Problem 6.1: Prime Numbers in Range (Medium)
Write a program to print all prime numbers between two given numbers (inclusive).

Example: Between 10 and 30: 11, 13, 17, 19, 23, 29

Hint: Outer loop for each number in range. For each number, use inner loop to check if it's prime (like Problem 2.5).

"""

for num in range(100):
    if num > 0:
        is_prime = True
        
        for i in range(2,num):
            if num % i == 0:
                is_prime = False
                break
            
        if is_prime:
            print(num, end= " ")
```

```python
"""

Problem 6.2: GCD Calculator (Medium)
Write a program to find the Greatest Common Divisor (GCD) of two numbers using the Euclidean algorithm.

Algorithm:

Divide the larger number by the smaller number
Replace the larger number with the smaller number
Replace the smaller number with the remainder
Repeat until the remainder is 0
The GCD is the last non-zero remainder
Hint: While b != 0: store remainder (a%b), then a=b, b=remainder. When loop ends, a is the GCD.

"""

def gcd(a, b):
    while b != 0:
        remainder = a % b
        a = b
        b = remainder
    return a

num1 = int(input("Enter larger number: "))
num2 = int(input("Enter smaller number: "))

result = gcd(num1, num2)
print("The GCD is:", result)
```

```python
"""

Problem 6.3: Perfect Number Checker (Hard)
Write a program to check if a number is a perfect number.

A perfect number is a positive integer that is equal to the sum of its proper positive divisors (excluding the number itself).

Example: 6 is perfect because 6 = 1 + 2 + 3

Example: 28 is perfect because 28 = 1 + 2 + 4 + 7 + 14

Hint: Loop from 1 to n-1. If i divides n (n%i==0), add i to sum. Compare sum with n.

"""

number = int(input("Enter number: "))
total = 0

for i in range(1, number + 1):
    if number % i == 0:
        total += i
        
if total == number:
    print(f"{number} is Perfect Number.")
else: 
    print(f"{number} is Not Perfect Number.")  
```

```python
"""

Problem 6.4: LCM Calculator (Medium)
Write a program to find the Least Common Multiple (LCM) of two numbers.

Hint: LCM(a, b) = (a × b) / GCD(a, b)

You can use the GCD function from Problem 6.2.

Hint: First calculate GCD using Euclidean algorithm (Problem 6.2), then use formula: LCM = (a * b) // GCD.

"""

def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a 

# LCM Calculator
def lcm(num1, num2):
    result = (num1 * num2) / gcd(num1, num2)
    return result

print(lcm(12, 18))
```

```python
"""

Problem 6.5: Number to Words (Hard)
Write a program that converts a number (0-999) to words.

Examples:

0: Zero
25: Twenty Five
123: One Hundred Twenty Three
500: Five Hundred
Hint: Create lists for ones, teens, tens, etc. Extract hundreds, tens, ones digits. Convert each part to words and combine.

"""

def number_to_words(num):
    ones = ["Zero","One","Two","Three","Four","Five","Six","Seven","Eight","Nine"]
    teens = ["Ten","Eleven","Twelve","Thirteen","Fourteen","Fifteen","Sixteen",
             "Seventeen","Eighteen","Nineteen"]
    tens = ["","","Twenty","Thirty","Forty","Fifty","Sixty","Seventy","Eighty","Ninety"]    

    if num == 0:
        return "Zero"

    words = ""

    # Hundreds place
    if num >= 100:
        words += ones[num // 100] + " Hundred"
        num = num % 100
        if num > 0:
            words += " "

    # Tens and ones
    if 10 <= num <= 19:
        words += teens[num - 10]
    else:
        if num >= 20:
            words += tens[num // 10]   # FIXED
            num = num % 10
            if num > 0:
                words += " "
        if 0 < num < 10:
            words += ones[num]

    return words

# Example
print(number_to_words(0))
print(number_to_words(25))
print(number_to_words(123))
print(number_to_words(500))
print(number_to_words(699))
```

```python
"""

Problem 6.6: Strong Number Checker (Hard)
Write a program to check if a number is a Strong number.

A Strong number is a number whose sum of factorial of digits is equal to the original number.

Example: 145 is a strong number because 1! + 4! + 5! = 1 + 24 + 120 = 145

Hint: Extract each digit, calculate its factorial, add to sum. Compare sum with original number.

"""

# Strong Number Checker

num = int(input("Enter a number: "))
original = num
sum_factorial = 0

while num > 0:
    digit = num % 10
    
    # calculate factorial of digit
    fact = 1
    for i in range(1, digit + 1):
        fact = fact * i
    
    sum_factorial += fact
    num = num // 10

# check strong number
if sum_factorial == original:
    print(original, "is a Strong number")
else:
    print(original, "is not a Strong number")
```

```python
"""

Problem 6.7: Pascal's Triangle (Hard)
Write a program to print Pascal's Triangle up to n rows.

Example for n=5:

    1
   1 1
  1 2 1
 1 3 3 1
1 4 6 4 1
Each number is the sum of the two numbers directly above it.

Hint: Use formula: C(n,k) = C(n,k-1) * (n-k+1) / k, where C(n,0)=1. Or use nested loops with list to store previous row.

"""

def pascal_triangle(n):
    for i in range(n):
        # spacing for triangle shape
        print(" " * (n - i), end="")

        val = 1
        for j in range(i + 1):
            print(val, end=" ")
            val = val * (i - j) // (j + 1)

        print()

# Example
pascal_triangle(5)
```

```python
"""

Problem 6.8: Simple Calculator (Medium)
Create a simple calculator program that:

Shows a menu of operations: Add, Subtract, Multiply, Divide, Exit
Takes two numbers as input
Performs the selected operation
Shows the result
Repeats until the user chooses to exit
Handles division by zero error
Hint: Use while loop. Display menu, get choice. Use if-elif to perform operation. Check if divisor is 0 before division.

"""

def calculator():
    while True:
        print("\nSimple Calculator")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == "5":
            print("Exiting calculator...")
            break

        if choice in ["1", "2", "3", "4"]:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))

            if choice == "1":
                print("Result:", num1 + num2)

            elif choice == "2":
                print("Result:", num1 - num2)

            elif choice == "3":
                print("Result:", num1 * num2)

            elif choice == "4":
                if num2 == 0:
                    print("Error: Division by zero is not allowed.")
                else:
                    print("Result:", num1 / num2)

        else:
            print("Invalid choice. Please select 1-5.")

# Run the calculator
calculator()
```
