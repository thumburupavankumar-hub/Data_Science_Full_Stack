```python
"""

Problem 7.1: Collatz Conjecture (Medium)
The Collatz conjecture is one of the most famous unsolved problems in mathematics. The rules are:

Start with any positive integer n
If n is even, divide it by 2
If n is odd, multiply it by 3 and add 1
Repeat until you reach 1
Write a program that:

Takes a number as input
Shows the complete sequence until it reaches 1
Counts how many steps it took
Example: Starting with 6: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 (8 steps)

Hint: While n != 1: if n%2==0, n=n//2, else n=3*n+1. Count steps and print each value.

"""

def collatz(n):
    steps = 0

    print("Sequence:")

    while n != 1:
        print(n, end=" → ")

        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1

        steps += 1

    print(1)  # print the final 1
    print("Total steps:", steps)


# Take input
num = int(input("Enter a positive integer: "))

collatz(num)
```

```python
"""

Problem 7.2: Caesar Cipher (Hard)
Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3

'A' becomes 'D'
'Hello' becomes 'Khoor'
Your program should:

Ask for a message and shift value
Provide options to encrypt or decrypt
Handle both uppercase and lowercase letters
Keep non-alphabetic characters unchanged
Hint: Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

"""

def caesar_cipher(text, shift, mode):
    result = ""

    if mode == "decrypt":
        shift = -shift

    for char in text:
        if char.isalpha():
            if char.isupper():
                base = ord('A')
            else:
                base = ord('a')

            # shift within alphabet range
            new_char = chr((ord(char) - base + shift) % 26 + base)
            result += new_char
        else:
            result += char

    return result


while True:
    print("\nCaesar Cipher")
    print("1. Encrypt")
    print("2. Decrypt")
    print("3. Exit")

    choice = input("Choose option: ")

    if choice == "3":
        print("Goodbye!")
        break

    message = input("Enter message: ")
    shift = int(input("Enter shift value: "))

    if choice == "1":
        encrypted = caesar_cipher(message, shift, "encrypt")
        print("Encrypted message:", encrypted)

    elif choice == "2":
        decrypted = caesar_cipher(message, shift, "decrypt")
        print("Decrypted message:", decrypted)

    else:
        print("Invalid choice.")
```

```python
"""

Problem 7.3: ATM Machine Simulator (Hard)
Create an ATM machine simulator with the following features:

Initial balance: 10000
PIN: 1234 (3 attempts allowed)
Menu options:
Check Balance
Withdraw Money (check for sufficient balance and valid amount)
Deposit Money
Change PIN
Exit
Keep transaction history (last 5 transactions)
Input validation for all operations
Hint: Use variables for balance, pin. Loop for PIN attempts. Use while loop for menu. Use list to store transaction history. Validate amounts > 0 and balance >= withdrawal.

"""

balance = 10000
pin = "1234"
transactions = []

# PIN authentication
attempts = 3
while attempts > 0:
    entered_pin = input("Enter PIN: ")
    
    if entered_pin == pin:
        print("Login successful!")
        break
    else:
        attempts -= 1
        print("Incorrect PIN. Attempts left:", attempts)

if attempts == 0:
    print("Too many incorrect attempts. Card blocked.")
    exit()


while True:
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Withdraw Money")
    print("3. Deposit Money")
    print("4. Change PIN")
    print("5. Transaction History")
    print("6. Exit")

    choice = input("Select option: ")

    if choice == "1":
        print("Current Balance:", balance)

    elif choice == "2":
        amount = float(input("Enter withdrawal amount: "))

        if amount <= 0:
            print("Invalid amount.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print("Withdrawal successful.")
            transactions.append(f"Withdraw: {amount}")

    elif choice == "3":
        amount = float(input("Enter deposit amount: "))

        if amount <= 0:
            print("Invalid amount.")
        else:
            balance += amount
            print("Deposit successful.")
            transactions.append(f"Deposit: {amount}")

    elif choice == "4":
        new_pin = input("Enter new PIN: ")
        confirm_pin = input("Confirm new PIN: ")

        if new_pin == confirm_pin:
            pin = new_pin
            print("PIN changed successfully.")
        else:
            print("PIN mismatch.")

    elif choice == "5":
        print("Last Transactions:")
        for t in transactions[-5:]:
            print(t)

    elif choice == "6":
        print("Thank you for using the ATM.")
        break

    else:
        print("Invalid option.")
```
